# Crop Disease Detection - Exploratory Data Analysis & Preprocessing

This notebook covers initial dataset inspection, feature analysis, data distribution checks, and image preprocessing steps for the **Crop Disease Detection** system.

## 1. Environment & Library Setup

```python
import os
import warnings
warnings.filterwarnings('ignore')

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
import sys
sys.path.append('..')
import config
```

## 2. Dataset Overview & Target Class Verification

```python
print(f"Total Number of Target Disease Classes: {config.NUM_CLASSES}")
print("Sample Class Labels:")
for idx, name in enumerate(config.CLASS_NAMES[:10]):
    print(f"  {idx+1}. {name}")
```

## 3. Class Distribution Analysis

```python
# Visualize class distribution if dataset directory exists locally
train_dir = config.DATASET_DIR / 'train'
if train_dir.exists():
    counts = {}
    for class_folder in os.listdir(train_dir):
        folder_path = train_dir / class_folder
        if folder_path.is_dir():
            counts[class_folder] = len(os.listdir(folder_path))
    
    df_counts = pd.DataFrame(list(counts.items()), columns=['Class', 'Image_Count']).sort_values('Image_Count', ascending=False)
    
    plt.figure(figsize=(14, 8))
    sns.barplot(data=df_counts, y='Class', x='Image_Count', palette='viridis')
    plt.title('Distribution of Images per Crop Disease Class')
    plt.xlabel('Number of Images')
    plt.ylabel('Crop Disease Class')
    plt.tight_layout()
    plt.show()
else:
    print("Dataset directory not found locally. Follow data/README.md to download PlantVillage dataset.")
```

## 4. Image Preprocessing & OpenCV Visualization

```python
def preprocess_and_display(image_path):
    """Reads, resizes, and normalizes a leaf image using OpenCV for visualization."""
    img_bgr = cv2.imread(str(image_path))
    if img_bgr is None:
        print(f"Could not load image at {image_path}")
        return
    
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img_resized = cv2.resize(img_rgb, config.IMG_SIZE)
    img_normalized = img_resized / 255.0
    
    fig, axes = plt.subplots(1, 3, figsize=(12, 4))
    axes[0].imshow(img_rgb)
    axes[0].set_title("Original RGB Image")
    axes[0].axis('off')
    
    axes[1].imshow(img_resized)
    axes[1].set_title(f"Resized ({config.IMG_WIDTH}x{config.IMG_HEIGHT})")
    axes[1].axis('off')
    
    axes[2].imshow(img_normalized)
    axes[2].set_title("Normalized [0.0 - 1.0]")
    axes[2].axis('off')
    
    plt.tight_layout()
    plt.show()

print("Preprocessing helper function defined.")
```
